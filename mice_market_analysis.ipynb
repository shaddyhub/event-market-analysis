# 🎯 MICE & Corporate Events Market Analysis
> *Market sizing, city benchmarking, and client satisfaction drivers in the B2B events industry.*

**Dataset:** 200 events | 6 event types | 10 destination cities | 3 years  
**Tools:** Python · Pandas · Matplotlib · Seaborn  
**Business Question:** *Which destinations, event types, and deal sizes are most valuable?*

---

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (11, 5)

df = pd.read_csv('data/mice_events_sample.csv')
print(f"✅ Loaded {len(df)} events across {df['city'].nunique()} cities")
print(df.describe()[['delegates','total_value','satisfaction_score']].round(1))
```

```python
# Market overview
print("=" * 50)
print("       MICE MARKET OVERVIEW")
print("=" * 50)
print(f"  Total Events Analysed:    {len(df)}")
print(f"  Total Market Value:       €{df['total_value'].sum()/1e6:.1f}M")
print(f"  Avg Event Value:          €{df['total_value'].mean():,.0f}")
print(f"  Avg Delegates per Event:  {df['delegates'].mean():.0f}")
print(f"  Avg Satisfaction Score:   {df['satisfaction_score'].mean():.1f}/5.0")
print("=" * 50)
```

## 🏙️ Top Destinations by Market Value

```python
city_stats = df.groupby('city').agg(
    total_value=('total_value','sum'),
    avg_value=('total_value','mean'),
    event_count=('event_type','count'),
    avg_satisfaction=('satisfaction_score','mean')
).sort_values('total_value', ascending=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Total value
axes[0].barh(city_stats.index, city_stats['total_value']/1000,
             color=sns.color_palette("viridis", len(city_stats)))
axes[0].set_title('Total Market Value by City (€K)', fontweight='bold')
axes[0].set_xlabel('Total Value (€K)')

# Satisfaction
colors_sat = ['#2ecc71' if v >= 4.2 else '#f39c12' if v >= 3.8 else '#e74c3c'
              for v in city_stats['avg_satisfaction']]
axes[1].barh(city_stats.index, city_stats['avg_satisfaction'], color=colors_sat)
axes[1].axvline(4.0, color='navy', linestyle='--', alpha=0.6, label='4.0 benchmark')
axes[1].set_title('Avg Client Satisfaction by City', fontweight='bold')
axes[1].set_xlabel('Satisfaction Score (/5)')
axes[1].legend()

plt.tight_layout()
plt.savefig('city_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 📊 Event Type Mix

```python
# Event type breakdown
event_stats = df.groupby('event_type').agg(
    total_value=('total_value','sum'),
    avg_delegates=('delegates','mean'),
    avg_pppd=('price_per_person_per_day','mean')
).sort_values('total_value', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14,5))

# Pie chart of value mix
wedge_props = {'edgecolor':'white','linewidth':2}
axes[0].pie(event_stats['total_value'], labels=event_stats.index,
            autopct='%1.1f%%', colors=sns.color_palette("Set2", len(event_stats)),
            wedgeprops=wedge_props, startangle=90)
axes[0].set_title('Market Value Share by Event Type', fontweight='bold')

# Avg delegates
axes[1].bar(event_stats.index, event_stats['avg_delegates'],
            color=sns.color_palette("muted", len(event_stats)))
axes[1].set_title('Avg Delegates by Event Type', fontweight='bold')
axes[1].set_ylabel('Avg Delegates')
plt.xticks(rotation=20)

plt.tight_layout()
plt.savefig('event_type_mix.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 📈 Year-on-Year Growth

```python
yoy = df.groupby('year').agg(
    total_value=('total_value','sum'),
    event_count=('event_type','count'),
    avg_satisfaction=('satisfaction_score','mean')
).round(2)

fig, axes = plt.subplots(1, 2, figsize=(12,4))

axes[0].bar(yoy.index.astype(str), yoy['total_value']/1e6,
            color=['#3498db','#2ecc71','#e74c3c'])
axes[0].set_title('Total Market Value YoY (€M)', fontweight='bold')
axes[0].set_ylabel('Value (€M)')

axes[1].plot(yoy.index, yoy['avg_satisfaction'], 'o-', color='#9b59b6',
             linewidth=2.5, markersize=10)
axes[1].set_ylim(3.5, 5.0)
axes[1].set_title('Client Satisfaction Score YoY', fontweight='bold')
axes[1].set_ylabel('Score (/5)')

plt.tight_layout()
plt.savefig('yoy_trends.png', dpi=150, bbox_inches='tight')
plt.show()
```
